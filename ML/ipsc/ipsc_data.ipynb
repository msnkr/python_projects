```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("match_results.csv", parse_dates=["Date"])
```

```python
data["Month"] = data["Date"].dt.month
data["Day"] = data["Date"].dt.day
data["Year"] = data["Date"].dt.year
```

```python
data.drop(["Category", "Date"], axis=1, inplace=True)
```

```python
sns.scatterplot(data=data, x="Match Percentage", y="Match Points", hue="Day")
```
