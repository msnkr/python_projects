```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.rcParams["figure.figsize"] = (14, 6)
```

```python
data = pd.read_csv("match_results.csv", parse_dates=["Date"])
```

```python
data["Month"] = data["Date"].dt.month
data["Day"] = data["Date"].dt.day
data["Year"] = data["Date"].dt.year
```

```python
data.drop(["Category"], axis=1, inplace=True)
```

```python
data.dtypes
```

```python
data.head()
```

```python
sns.scatterplot(data=data, x="Match Percentage", y="Match Points", hue="Month")
```

```python
dez = data[data["Name"] == "Coetzee, Desmond"]
```

```python
dez
```

```python
sns.lineplot(data=dez, x="Date", y="Match Percentage", )
plt.xticks(rotation=45)
plt.axhline(data["Match Percentage"].mean(),
            color="red", linestyle="--", label="Mean")
```

```python
sns.lineplot(data=dez, x='Month', y="Match Points")
plt.xticks(rotation=45)
plt.axhline(data["Match Points"].mean(), color="red",
            linestyle="--", label="Mean")
```
